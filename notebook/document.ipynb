```python
## Create a simple txt file

import os
os.makedirs("../data/text_files",exist_ok=True)
```

```python
sample_texts={
    "../data/text_files/sample1.txt":"""Machine Leaning Introduction
    Machine learning is a branch of artificial intelligence that allows computers to learn from data and make predictions without being explicitly programmed for every single task.

    Main Types of Machine Learning:

    Supervised Learning: Trains a model on labeled data, like matching emails to "spam" or "not spam".
    Unsupervised Learning: Finds hidden structures or groupings in unlabeled data, such as clustering customer segments.
    Reinforcement Learning: Learns optimal actions through trial and error to maximize a reward signal.

    """

}

for filepath,content in sample_texts.items():
    with open(filepath,"w",encoding="utf-8") as f:
        f.write(content)

print(" SAMPLE TEXT FILES CREATED")        
```

```python
### TextLoader
from langchain_community.document_loaders import TextLoader

loader = TextLoader("../data/text_files/ML.txt",encoding="utf-8")
document=loader.load()
print(document)
```

```python
### Directory Folder

from langchain_community.document_loaders import DirectoryLoader, TextLoader

# Initialize the DirectoryLoader with valid parameters
dir_loader = DirectoryLoader(
    path="../data/text_files", 
    glob="**/*.txt",                     # Pattern to match files
    loader_cls=TextLoader,               # Loader class to use
    loader_kwargs={"encoding": "utf-8"}, # Keyword arguments passed to TextLoader
    show_progress=False,                 # Show progress bar
    use_multithreading=True,             # Use multithreading
    max_concurrency=4                    # Number of concurrent threads
)

documents = dir_loader.load()

print(documents[0])
```

```python
from langchain_community.document_loaders import DirectoryLoader, PyMuPDFLoader

# Initialize the DirectoryLoader for PDFs
dir_loader = DirectoryLoader(
    path="../data/pdf", 
    glob="**/*.pdf", 
    loader_cls=PyMuPDFLoader,
    show_progress=False, 
    use_multithreading=True, 
    max_concurrency=4 
)

pdf_documents = dir_loader.load()
pdf_documents
```
